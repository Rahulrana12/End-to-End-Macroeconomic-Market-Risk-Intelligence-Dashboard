```python
import pandas as pd
import numpy as np
import kagglehub
import psycopg2
path = kagglehub.dataset_download("belbino/us-tariff-and-trade-war-impact-dataset-2018-present")
print("Path to dataset files:", path)
```

```python
tariff_rates = pd.read_csv(f"{path}/tariff_rates.csv")
market_reaction = pd.read_csv(f"{path}/market_reaction.csv")
trade_balance = pd.read_csv(f"{path}/trade_balance.csv")
news_headlines = pd.read_csv(f"{path}/tariff_news_headlines.csv")
```

```python
for df in [tariff_rates, market_reaction, trade_balance, news_headlines]:
 df.columns = df.columns.str.lower().str.strip().str.replace(" ","_").str.replace("[^a-zA-Z0-9_]","",regex=True)
```

```python
market_reaction = market_reaction.sort_values("date")
```

```python
market_reaction["shanghai_composite"] = market_reaction["shanghai_composite"].ffill()
market_reaction["shanghai_composite"] = market_reaction["shanghai_composite"].bfill()
```

```python
tariff_rates["url"] = tariff_rates["url"].fillna("Source Unspecified")
```

```python
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

url = URL.create(
    drivername="postgresql+psycopg2",
    username="postgres",
    password="Rahul@rana7217",
    host="localhost",
    port=5432,
    database="market_intelligence_dashboard",
)

engine = create_engine(url)

with engine.connect() as conn:
    print("Connected successfully!")
```

```python
tariff_rates.to_sql('tariff_rates', engine, if_exists='replace', index=False)
market_reaction.to_sql('market_reaction', engine, if_exists='replace', index=False)
trade_balance.to_sql('trade_balance', engine, if_exists='replace', index=False)
news_headlines.to_sql('news_headlines', engine, if_exists='replace', index=False)

print("Data successfully loaded into PostgreSQL!")
```
